```python
import networkx as nx
G = nx.Graph()
```

```python
G.add_node(1)
```

```python
G.add_nodes_from([2, 3])
import matplotlib.pyplot as plt
```

```python
subax1 = plt.subplot()
nx.draw(G, with_labels=True, font_weight='bold')
```

```python
gp_node = nx.DiGraph()
node = [("GP",{"color":"green"}),("ZMH",{"color":"red"})]
gp_node.add_nodes_from(node)

subax2 = subax1.twinx()
pos = nx.spring_layout(gp_node)
pos["GP"] = (0,1) # force main node at center
pos["ZMH"] = (0, 0)
nx.draw(gp_node,pos,with_labels=True, font_weight='bold', font_size=10, node_size=1000)
print(pos)
```
